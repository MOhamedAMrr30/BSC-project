# Import header files

```python
# import PyPore3D wrappers that will be used in the current notebook
import pypore3d
from pypore3d import *
from pypore3d.p3dFiltPy import *
from pypore3d.p3dSkelPy import *
from pypore3d.p3dBlobPy import *
from pypore3d.p3dSITKPy import *
from pypore3d.p3dSITKPy_16 import *

# import external python libraries
import time
```

# Set common variables that will be used in this notebook

```python
# change the folder path to the location of the Input/output data
data_input_folder = 'Data/Input/'
data_output_folder = 'Data/Output/'

# volume dimensions
x = 300
y = 300
z = 300

# other variables

#SC1
file_tag = 'test'
voxel_size = 0.0026

## TR2
#file_tag = 'TR2'
#voxel_size = 0.00137
 
## FL1
#file_tag = 'FL1'
#voxel_size = 0.00137
 
## SSK36
#file_tag = 'SSK36'
#voxel_size = 0.0018
```

# Read the file of the original data

```python
inFile = data_input_folder+file_tag+"_300x300x300.raw"
print(inFile)
inImg = py_p3dReadRaw8(inFile,x,y,dimz = z)
```

# Apply filters
    - In this example, we apply Median, Autothresholding (Otsu threshold), Erosion, and Dilation
    - The data can be written in the end of the filter computation to save the file and investigate separately. 

```python
# Compute Median

start = time.time()
outImg = py_p3dMedianFilter8(inImg,x,y,dimz = z, width = 3 )

# Apply autothresholding (Otsu)

outImg= py_p3dAutoThresholding8(outImg, x,y,dimz = z,methodNum = 2)
#invert (if needed)
invert_vol(outImg, x,y,z)

# Erode, dilate
outImg = py_p3d_Erode(outImg, x, y, dimz=z, kWidth = 1)
outImg = py_p3d_Dilate(outImg, x, y, dimz=z, kWidth = 1)
end = time.time()
print("Filtering time: ", end - start) # print computation time if needed
####################

# write file (if needed, for example, to be for visualization or to be reused later to save computation time)
py_p3dWriteRaw8( outImg, data_output_folder+file_tag+'_median_otsu_eroded_dilated.raw',x,y,dimz = z) 
```

# Basic and anisotropy analysis

```python
#basic/anisotropy analysis

inFile = data_output_folder+file_tag+"_median_otsu_eroded_dilated.raw"
print (inFile) # python built-in function
outImg = py_p3dReadRaw8(inFile,x,y,dimz = z)

##########
start = time.time()
out_basic_stats = py_p3dBasicAnalysis(outImg, x,y,dimz = z, resolution = voxel_size)
out_anisotropy_stats = py_p3dAnisotropyAnalysis (outImg, x,y,dimz = z, resolution = voxel_size)
end = time.time()
print("Basic/Anisotropy analaysis time : ", end - start)
##########

print(out_basic_stats.Cv, ",", out_basic_stats.Mv, ",", out_basic_stats.Sv, ",", out_basic_stats.Vv, "\n")
print(out_anisotropy_stats.E, ",", out_anisotropy_stats.I, "\n")
```

# Connected component (Blob) analysis

```python
#blob analysis

inFile = data_output_folder+file_tag+"_median_otsu_eroded_dilated.raw"
print (inFile)
outImg = py_p3dReadRaw8(inFile,x,y,dimz=z)

##########
start = time.time()
stat_file = data_output_folder+file_tag+"_blob_test.txt"
out_blob_stat =  py_p3dBlobAnalysis( outImg, x,y,dimz=z, resolution = voxel_size, conn = 26,blob_analysis_file=stat_file)
end = time.time()
print("Runtime of Blob Analysis is {end - start}", end - start)
##########

```

# Skeletonization

```python
#skeletonization

inFile = data_output_folder+file_tag+"_median_otsu_eroded_dilated.raw"
print (inFile)
outImg = py_p3dReadRaw8(inFile,x,y,dimz=z)

skel_thresh = 5

####################
start = time.time()
sklImg = py_p3dLKCSkeletonization(outImg, x,y,dimz = z)
sklImg = py_p3dSkeletonPruning(sklImg, x,y,dimz = z,thresh = skel_thresh)
sklImg =py_p3dSkeletonLabeling(sklImg,x,y,dimz = z)
end = time.time()
print("Skeletonization/pruning/labeling time: ", end - start)
####################

outFile = data_output_folder+file_tag+'_median_otsu_skeleton_LKC_pruned_labeled.raw'
print (outFile)
py_p3dWriteRaw8(sklImg,outFile ,x,y,dimz = z) # to be used in skeleton analysis
```

# Skeleton Analysis
Note: This module takes long computation time may be up to hours (depending on the workstation it runs on)

```python
inFile = data_output_folder+file_tag+"_median_otsu_eroded_dilated.raw"
outImg = py_p3dReadRaw8(inFile,x,y,dimz=z)

sklFile = data_output_folder+file_tag+"_median_otsu_skeleton_LKC_pruned_labeled.raw"
sklImg = py_p3dReadRaw8(inFile,x,y,dimz=z)

# skeleton analysis
PSkeletonStats = py_p3dSkeletonAnalysis(outImg, sklImg,x,y,dimz=z, resolution = voxel_size)
print ("\n PSkeletonStats.ConnectivityDensity:",PSkeletonStats.ConnectivityDensity,"\n")  
```
